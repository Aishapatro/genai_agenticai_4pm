# Loop through a Dictionary 

```python
d={'Name':'Aisha','ID':234568,'DOB':2005,'job':'Data Science','Address':'HYD'}
d
```

```python
for i in d:
    print(i,':',d[i])
```

```python
for i in d:
    print(d[i])
```

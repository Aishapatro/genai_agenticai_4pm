# Hallow Pyramid Pattern 

```python
for i in range(1,6):
    for j in range(5-i):
        print(' ',end=' ')
    for j in range(2 * i - 1):
        if j == 0 or j == 2 * i - 2 or i == 5:
            print('*',end=' ')
        else:
            print(' ',end=' ')
    print()
```

# Hallow Number Pyramid 

```python
n=5
for i in range(1, n + 1):
    for j in range(n - i):
        print(' ',end= ' ')

    for j in range(1,2 * i):
        if j == 1 or j == 2 * i - 1 or i == n:
            print(i, end=' ')
        else:
            print(' ', end=' ')
    print()
            
```

# Full Star Pyramid 

```python
n=5

for i in range(1, n + 1):

    for j in range(n - i):
        print(' ', end= ' ')

    for j in range(2 * i - 1):
        print('*', end= ' ')

    print()
```

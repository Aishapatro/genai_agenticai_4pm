# Set Operation 

## Union 

```python
a={10,20,30,40,50,60}
b={50,60,70,80,90}
c={90,100,110}
```

```python
a | b
```

```python
a.union(b)
```

```python
a.union(b,c)
```

```python
a.update(b,c)
a
```

## Intersection 

```python
a={1,2,3,4,5}
b={4,5,6,7,8}
```

```python
a & b
```

```python
a.intersection(b)  intersection of a & b
```

```python
a.intersection_update(b)
a
```

```python
b.intersection_update(a)
b
```

## Difference 

```python
a={1,2,3,4,5}
b={4,5,6,7,8}
```

```python
a - b
```

```python
b.difference(a)
```

```python
b.difference_update(a)
b
```

```python
a.difference_update(b)
a
```

## Symmetric Difference 

```python
a1={12,13,14,15,16,17}
b1={16,17,18,19,20,21}
```

```python
a1 ^ b1
```

```python
a1.symmetric_difference(b1)
```

```python
a1.symmetric_difference(b1)
a1
```

```python
b1.symmetric_difference(a1)
b1
```
